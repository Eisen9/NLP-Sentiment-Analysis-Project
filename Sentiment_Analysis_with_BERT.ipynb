### Step 2: Load the Pre-trained Model and Tokenizer
We will load the BERT model and tokenizer from Hugging Face's model repository.

```
from transformers import BertTokenizer, BertForSequenceClassification
from transformers import pipeline

# Load pre-trained BERT model and tokenizer
tokenizer = BertTokenizer.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')
model = BertForSequenceClassification.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')

# Create a sentiment analysis pipeline
sentiment_analysis = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer)
```

### Step 3: Define the Sentiment Analysis Function
This function processes the input text and provides a detailed sentiment analysis with confidence levels.

```
# Function to get more detailed sentiment analysis
def analyze_sentiment(text):
    result = sentiment_analysis(text)[0]
    
    label = result['label']
    confidence = round(result['score'] * 100, 2)
    
    # Determine sentiment and intensity level based on the label
    if "5" in label:
        sentiment = "positive"
        intensity = "High Positivity: 5/5"
    elif "4" in label:
        sentiment = "positive"
        intensity = "Moderate Positivity: 4/5"
    elif "3" in label:
        sentiment = "neutral"
        intensity = "Neutral: 3/5"
    elif "2" in label:
        sentiment = "negative"
        intensity = "Moderate Negativity: 2/5"
    elif "1" in label:
        sentiment = "negative"
        intensity = "High Negativity: 1/5"
    
    return f"Sentiment: {sentiment}, {intensity}, Confidence level: {confidence}%"
```

### Step 4: Test the Sentiment Analysis
You can now enter your own review in the text variable and run the function to see the result.

```
# Test the function with your own review
review = "This is awesome"
print(analyze_sentiment(review))
```
